### Using Pandas to Read and Manipulate Data

Reading data into Python natively can be a pain, especially if you have large datasets with different data types. Luckily, there are many packages like `pandas` and `numpy` that can help you read in data and manipulate it easily.

`pandas` is a package which has become increasingly popular in data science recently due to its easy-to-use and intuitive `DataFrames`.
Pandas `DataFrame`s are objects that are akin to fancy dictionaries (`dict`s).

```python
# using pd as an alias for pandas is a common convention
import pandas as pd

# We need to get the filename + extension (.csv, .txt, .dat, etc.)
filename = "<PLACE NAME OF FILE HERE>"

# Strings can be denoted by either ' or "
file_path = 'data/' + filename

# Often pandas dataframes have the alias df
nea_df = pd.read_csv(file_path, sep=',', comment='#') # sep tells pandas how to split the data, ',' means split by commas

print("Type:", type(nea_df))

# We can print the column names and first few rows using the .head() method
# It's a useful way to get a quick look at the data and make sure it was read in correctly
print(nea_df.head())
```

You might have noticed when we printed the `head` of the NEA data that there are numerous columns with shorhand names that are not immediately obvious. The NEA data follows a specific organization structure, which is described in the [NEA Data Column Descriptions](https://exoplanetarchive.ipac.caltech.edu/docs/API_PS_columns.html) and at the top of the downloaded data file. Each column has a name, description, and unit. As a starting point for understanding the column names, any data column that starts with `pl_` is a planetary parameter, `st_` is a stellar parameter, and `sy_` is a system parameter. For example, the column `pl_rade` is the planetary radius in units of Earth radii, while `st_teff` is the stellar effective temperature in Kelvin.

Let's see how to access specific columns and rows of the NEA data using `pandas`.

> INDEXING
>
> In Python, indexing (denoted by brackets) starts at 0. So the first item of a dataset is index 0, the second item is index 1, and so on. If you have a list `my_list = [10, 20, 30]`, then `my_list[0]` will return `10`, `my_list[1]` will return `20`, and `my_list[2]` will return `30`. If you try to access an index that is out of range, such as `my_list[3]`, you will get an error.
>
> This is a common source of confusion for new programmers, so be careful when indexing into data!

```python
# We can access columns by name
print(nea_df['pl_rade'], '\n') # \n is a newline character, just to make the output easier to read

# Or you can access rows by index using .iloc[row_index]
print(nea_df.iloc[0])
```

AWESOME! We now have our exoplanet data in a `pandas` `DataFrame` and we can start manipulating it. Now, what if we only want to look at planets that are smaller than 2 Earth radii? We can use a conditional statement to filter the data and view only the planets that meet our criteria.

In Python, you can check whether things are greater than (`>`), less than (`<`), equal to (`==`), not equal to (`!=`), greater than or equal to (`>=`), or less than or equal to (`<=`) a certain value. For a list or array, this will give the location in the list where this condition is true or false (called a boolean array). Let's use this to find our terrestrial planets!

```python
# Get the index of the rows where the planetary radius is less than 2 Earth radii
indices = nea_df['pl_rade'] < 2

# Use boolean array to filter
filtered_df = nea_df[indices]

# Print the filtered dataframe
print(filtered_df['pl_rade'].head())
```
