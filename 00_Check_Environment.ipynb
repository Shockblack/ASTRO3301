If you have any issues with the environment setup, please let me know! You can also install packages individually using `pip install <package_name>` or `conda install <package_name>` if you prefer (I prefer using `pip`).

````python
# Check Python version
import sys

error_tracker = False

if sys.version_info < (3, 14):
    print("-"*75)
    print(f"You are using Python {sys.version_info.major}.{sys.version_info.minor}. We cannot ensure all code will run properly on versions of Python below 3.11.")
    print(f"Please upgrade to Python 3.14 or higher.")
    print("-"*75)
    error_tracker = True

# Check to make sure all packages are installed and import them
packages = ['numpy', 'scipy', 'matplotlib', 'pandas', 'astropy', 'astroquery', 'jupyter', 'ipykernel']

for package in packages:
    try:
        __import__(package)
        print(f"{package} is installed.")
    except ImportError:
        print("-"*75)
        print(f"{package} is not installed. Please install it using pip or conda.\nRun:")
        print(f"pip install {package}")
        print("-"*75)
        error_tracker = True

print("\n")

if not error_tracker:
    print("All packages are installed. You are ready to start the bootcamp!")
else:
    print("Please install the missing packages and/or upgrade your Python version.")
````
