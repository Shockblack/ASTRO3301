## 1. Planet Size Distribution

Most exoplanets discovered to date were found using the transit method, which measures the dip in brightness of a star as a planet passes in front of it. The depth of the transit ($\delta$) is proportional to the square of the ratio of the planet's radius to the star's radius following
$$
\delta \propto \left(\frac{R_p}{R_*}\right)^2,
$$
where $R_p$ is the radius of the planet and $R_*$ is the radius of the star. This means that we can measure the radius of a planet if we know the radius of its host star.
The *Kepler* mission has discovered thousands of exoplanets, and we can use this data to study the distribution of planet sizes. In this section, we will plot a histogram of the planet radii to see if there are any interesting features in the distribution.

First, let's load in our data. You should know the drill by now :)

```python
import pandas as pd

# Filename and extension
filename = "<PLACE NAME OF FILE HERE>"

file_path = 'data/' + filename

nea_df = pd.read_csv(file_path, sep=',', comment='#')
```

Let's start by plotting a scatter plot of the planet radius vs. orbital period.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.scatter(nea_df['pl_orbper'], nea_df['pl_rade'], s=10, color='blue', alpha=0.5)

plt.xscale('log')

# Labels and limits
plt.xlabel('Orbital Period (days)')
plt.ylabel('Planet Radius (Earth radii)')

plt.show()
```

Hm, that's hard to gain anything from. Let's zoom in on the smaller planets and shorter orbital periods.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.scatter(nea_df['pl_orbper'], nea_df['pl_rade'], s=10, color='blue', alpha=0.5)

plt.xscale('log')

# Labels and limits
plt.xlabel('Orbital Period (days)')
plt.ylabel('Planet Radius (Earth radii)')
plt.xlim(0.1, 1e3)
plt.ylim(0, 4)

plt.show()
```

I think I might see something! Maybe instead of a scatter plot we should try a density plot to see if we can find any interesting features in the distribution of planet sizes.

```python
# Limit our data to the smaller planets and shorter orbital periods
small_planet_idx = (nea_df['pl_rade'] < 4) & (nea_df['pl_orbper'] < 1000)

# Create a new dataframe with only the small planets
small_planet_df = nea_df[small_planet_idx]

# Make the orbital period column logarithmic
small_planet_df['pl_orbper_log'] = np.log10(small_planet_df['pl_orbper'])
```

```python
# Plot a density plot of the small planets
plt.hexbin(small_planet_df['pl_orbper_log'], small_planet_df['pl_rade'], gridsize=20, cmap='Oranges', mincnt=1)

plt.colorbar(label='Number of Planets')

plt.xlabel('Log$_{10}$ Orbital Period (days)')
plt.ylabel('Planet Radius (Earth radii)')

plt.show()
```

Whoah! Do you see that?! It looks like there are two distinct populations of planets. I wonder if we can see this more clearly by plotting a 1D histogram of the planet radii instead of a 2D density plot. Let's try that next.

```python
# Using plt.hist
plt.hist(small_planet_df['pl_rade'], bins=30, color='blue', alpha=0.7, edgecolor='black')

plt.xlabel('Planet Radius (Earth radii)')
plt.ylabel('Number of Planets')
```

Whoah! That certainly looks interesting. What you are seeing is the "radius gap" (or "Fulton gap" after the paper which first reported it), which is a dearth of planets with radii between 1.5 and 2 Earth radii. This gap is thought to be caused by photoevaporation, where the intense radiation from the host star strips away the atmospheres of planets that are too close to their stars. This separates the population of planets into super-Earths (1-1.5 Earth radii) and sub-Neptunes (2-4 Earth radii).
